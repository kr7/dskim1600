# Scatterplot

```python
import matplotlib.pyplot as plt

verbrauch       = [ 4.5, 6.0, 6.5,  5.2, 5.3, 7.3, 7.1, 6.9]
geschwindigkeit = [ 60,  90,  100,  80,  70,  130, 115, 120]

plt.scatter(geschwindigkeit, verbrauch, marker='x', c=['r']*4+['b']*4)

plt.title('Verbrauch als Funktion der Geschwindigkeit', fontsize=14)
plt.xlabel('Geschwindigkeit (km/h)', fontsize=14)
plt.ylabel('Verbrauch (l/100km)', fontsize=14)
plt.show()
```

# Interaktive Widgets: TabBar

```python
from google.colab import widgets
import matplotlib.pyplot as plt

tb = widgets.TabBar(["Q1", "Q2", "Q3", "Q4"], location='start')

with tb.output_to(0):
  plt.bar([1,2,3], [50, 70, 60],
    tick_label=['Jan', 'Feb', 'März'],
    color=["b","g", "r"])
  plt.ylabel('Umsatz (Millionen EUR)')
  plt.show()

with tb.output_to(1):
  plt.bar([1,2,3], [80, 90, 100],
    tick_label=['April', 'Mai', 'Juni'],
    color=["b","g", "r"])
  plt.ylabel('Umsatz (Millionen EUR)')
  plt.show()

with tb.output_to(2):
  plt.bar([1,2,3], [95, 110, 90],
    tick_label=['Juli', 'Aug', 'Sept'],
    color=["b","g", "r"])
  plt.ylabel('Umsatz (Millionen EUR)')
  plt.show()

with tb.output_to(3):
  plt.bar([1,2,3], [85, 40, 92],
    tick_label=['Okt', 'Nov', 'Dez'],
    color=["b","g", "r"])
  plt.ylabel('Umsatz (Millionen EUR)')
  plt.show()
```

# Beispiel: Analyse von Kraftstoffverbrauchsdaten

**Aufgabe**

5. Betrachten Sie die Kraftstoffverbrauchsdaten unter http://www.biointelligence.hu/mi/fuel_data.txt und laden Sie diese Daten in eine geeignete Datenstruktur (zB. in ein Pandas-Dataframe oder eine eigene Datenstruktur).
Auf einem Scatterplot sollen Sie den Verbrauch als Funktion der durschschnittlichen Geschwindigkeit darstellen. Die Farbe soll den Verkehrsverhältnissen entsprechen und der Typ des Markers soll die Art des Kraftstoffs zeigen. Sie sollen zwei Diagramme auf zwei Blätter eines TabBars produzieren:
- das erste Diagramm soll ein Scatterplot darstellen, welche die Daten zeigt, die während den Fahrten auf der Autobahn aufgenommen worden sind, und
- das zweite Diagramm soll jene Daten zeigen, die während den Fahrten auf der Landstrasse aufgenommen worden sind.

```python
import pandas as pd
df = pd.read_csv('http://www.biointelligence.hu/mi/fuel_data.txt', sep='\t')
df.head()
```

```python
tb = widgets.TabBar(["Autobahn", "Landstrasse"], location='start')

with tb.output_to(0):
  # Zwei Sequenzen von numerischen Werten
  verbrauch_95plus = df[ (df["road"]=="highway") & (df["avg.cons."]< 10) & (df["fuel type"] == "95+") ]["avg.cons."]
  geschwindigkeit_95plus = df[ (df["road"]=="highway") & (df["avg.cons."]< 10) & (df["fuel type"] == "95+") ]["speed"]

  farben = []
  for v in df[ (df["road"]=="highway") & (df["avg.cons."]< 10) & (df["fuel type"] == "95+") ]["trafic"]:
    if v == "low":
      farben.append("b")
    elif v == "normal":
      farben.append("g")
    else:
      farben.append("r")

  verbrauch_95fs = df[ (df["road"]=="highway") & (df["avg.cons."]< 10) & (df["fuel type"] == "95FS") ]["avg.cons."]
  geschwindigkeit_95fs = df[ (df["road"]=="highway") & (df["avg.cons."]< 10) & (df["fuel type"] == "95FS") ]["speed"]

  farben_95fs = []
  for v in df[ (df["road"]=="highway") & (df["avg.cons."]< 10) & (df["fuel type"] == "95FS") ]["trafic"]:
    if v == "low":
      farben_95fs.append("b")
    elif v == "normal":
      farben_95fs.append("g")
    else:
      farben_95fs.append("r")

  plt.scatter(geschwindigkeit_95plus, verbrauch_95plus, marker="x", c = farben);
  plt.scatter(geschwindigkeit_95fs, verbrauch_95fs, marker="o", c = farben_95fs );


with tb.output_to(1):
  verbrauch_95plus = df[ (df["road"]=="standard") & (df["avg.cons."]< 10) & (df["fuel type"] == "95+") ]["avg.cons."]
  geschwindigkeit_95plus = df[ (df["road"]=="standard") & (df["avg.cons."]< 10) & (df["fuel type"] == "95+") ]["speed"]

  farben = []
  for v in df[ (df["road"]=="standard") & (df["avg.cons."]< 10) & (df["fuel type"] == "95+") ]["trafic"]:
    if v == "low":
      farben.append("b")
    elif v == "normal":
      farben.append("g")
    else:
      farben.append("r")

  verbrauch_95fs = df[ (df["road"]=="standard") & (df["avg.cons."]< 10) & (df["fuel type"] == "95FS") ]["avg.cons."]
  geschwindigkeit_95fs = df[ (df["road"]=="standard") & (df["avg.cons."]< 10) & (df["fuel type"] == "95FS") ]["speed"]

  farben_95fs = []
  for v in df[ (df["road"]=="standard") & (df["avg.cons."]< 10) & (df["fuel type"] == "95FS") ]["trafic"]:
    if v == "low":
      farben_95fs.append("b")
    elif v == "normal":
      farben_95fs.append("g")
    else:
      farben_95fs.append("r")

  plt.scatter(geschwindigkeit_95plus, verbrauch_95plus, marker="x", c = farben);
  plt.scatter(geschwindigkeit_95fs, verbrauch_95fs, marker="o", c = farben_95fs );
```

# AUPR

**Aufgabe**

6. Wir benutzen einen Klassifikator zur Abwanderungsprognose. Der Klassifikator produziert kontinuierliche Werte als Vorhersagen: je größer der Wert, desto wahrscheinlicher ist es, laut dem Klassifikator, dass der Kunde seinen Vertrag kündigen wird. Die Vorhersagen des Klassifikators sind in der Liste <i>y_pred</i>. Da wir historische Daten bei dieser Analyse benutzen, wissen wir, welche Kunden ihre Verträge tatsächlich gekündigt haben (1) und welche nicht (0), diese Information befindet sich in der Liste <i>y</i>. Sie sollen die Precision-Recall-Kurve des Klassifikators darstellen und AUPR (d.h. die Fläche unter der Kurve) berechnen.

```python
y_pred = [0.9, 0.1, 0.2, 0.6, 0.8, 0.7, 0.5, 0.4, 0.6, 0.7 ]
y      = [1  ,   0,   0,   1,   0,   1,   0,   0,   1,   0 ]

from sklearn.metrics import average_precision_score
average_precision = average_precision_score(y, y_pred)
```
