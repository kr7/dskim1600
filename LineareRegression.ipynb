8. Während Sie das Modell trainieren, Sie möchten den durchschnittlichen quadratischen Fehler, d.h. $ \frac{1}{n}\sum\limits_{i=1}^{n} ( \hat{y}_i - y_i)^2 $ mit dem Gradientenabstiegsverfahren minimieren. Wovon hängt der Fehler ab? Wieviele partielle Ableitungen gibt es?


Da die Daten gegeben sind, hängt der Fehler nur von den Modellparametern ab.

9. Geben Sie eine Funktion an, welche die Werte der partiellen Ableitungen der Fehlerfunktion bei gegebenen Modellparameterwerten $w^{(0)}, w^{(1)}, w^{(2)}, w^{(3)}$ berechnet. Ergänzen Sie den unteren Code!

```python
# X = Daten
# y = die tatsächlichen Preise der Wohnungen
# w0, w1, w2, w3 = Modellparameter
def partielle_ableitungen(X, y, w0, w1, w2, w3):
  # hier kommt Ihr Code
  w = np.array([w0, w1, w2, w3])
  y_pred = X@w # Wir berechnen die geschätzen Werte aller Immobilien mit Matrizenmultiplikation
  y_diff = y_pred - y
  n = len(X) # Die Anzahl der Instanzen (Zeilen in X)

  # partielle Ableitungen nach w0, w1, w2 und w3
  df_w0 = np.sum(2*X[:,0]*y_diff)/n
  df_w1 = np.sum(2*X[:,1]*y_diff)/n
  df_w2 = np.sum(2*X[:,2]*y_diff)/n
  df_w3 = np.sum(2*X[:,3]*y_diff)/n

  return df_w0, df_w1, df_w2, df_w3
```

10. Geben Sie den Code an, welcher die optimalen Modellparameter mit dem Gradientenabstiegsverfahren bestimmt!

```python
all_mse = []
w0 = 5
w1 = 16
w2 = 4
w3 = 2
eps = 1e-5
for i in range(100):
  df_w0, df_w1, df_w2, df_w3 = partielle_ableitungen(X, y, w0, w1, w2, w3)
  w0 = w0 - eps*df_w0
  w1 = w1 - eps*df_w1
  w2 = w2 - eps*df_w2
  w3 = w3 - eps*df_w3

  # Fehler des Modells
  w = np.array([w0, w1, w2, w3])
  y_pred = X@w
  mse = np.mean((y_pred - y)**2)

  all_mse.append(mse)
```

```python
from matplotlib import pyplot as plt
plt.plot(all_mse)
plt.show()
```
